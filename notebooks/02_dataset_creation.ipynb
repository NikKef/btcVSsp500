```python
from pathlib import Path
import pandas as pd
```

```python
project_root = next(
    p for p in [Path.cwd(), *Path.cwd().parents]
    if (p / ".env").exists()
)
PROCESSED_DIR = project_root / "data" / "processed"
PROCESSED_DIR
```

```python
btc = pd.read_csv(
    PROCESSED_DIR / "btc_usd_daily.csv",
    index_col="date",
    parse_dates=True
)
```

```python
spy = pd.read_csv(
    PROCESSED_DIR / "spy_usd_daily.csv",
    index_col="date",
    parse_dates=True
)
spy = spy.rename(columns={"adjusted_close": "price_spy"})
```

```python
btc = btc[btc["volume"] > 0]
```

```python
btc["price_btc"] = (btc["low"] + btc["high"]) / 2.0
```

```python
combined = btc.merge(spy, on="date", how="inner")
combined = combined[["price_btc", "price_spy"]]
```

```python
combined.to_csv(
    PROCESSED_DIR / "btc_spy_daily.csv"
)

print("Full dataset:", len(combined))
print("Start:", combined.index.min())
print("End:", combined.index.max())
```
